Estructuras de Datos

```python
contactos = [
    {"id": 1, "nombre": "Marco", "telefono": "2871104293", "email": "marco@hotmail.com"},
    {"id": 2, "nombre": "Ana", "telefono": "2871122343", "email": "ana@hotmail.com"}
]

for contacto in contactos:
    print(contacto["nombre"])
```

```python
id_buscar = 2

for contacto in contactos:
    if contacto["id"] == id_buscar:
        print("Econtrado: ", contacto)
        break
```

```python
nombre = input("Nombre a buscar: ")

for contacto in contactos:
    if contacto["nombre"] == nombre:
        print("Encontrado: ", contacto["nombre"], contacto["telefono"])
        break
```
